```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Dataset .csv")
print(df.head())

top_cuisines = df['Cuisines'].dropna().str.split(', ').explode()

top_10 = top_cuisines.value_counts().head(10)
print(top_10)

top_10.plot(kind='bar')
plt.xlabel("Cuisines")
plt.ylabel("Count")
plt.title("Top 10 Cuisines")
plt.show()

percentage = (top_10 / top_cuisines.count()) * 100
print(percentage)

city_count = df['City'].value_counts()
print(city_count.head(10))

highest_restaurant_city = city_count.idxmax()
print(highest_restaurant_city)

avg_rating_city = df.groupby('City')['Aggregate rating'].mean()
print(avg_rating_city.sort_values(ascending=False).head(10))

highest_avg_rating_city = avg_rating_city.idxmax()
print(highest_avg_rating_city)

city_count.head(10).plot(kind='bar')
plt.xlabel("City")
plt.ylabel("Restaurants")
plt.title("Restaurants by City")
plt.show()

price_range = df['Price range'].value_counts().sort_index()
print(price_range)

price_range.plot(kind='bar')
plt.xlabel("Price Range")
plt.ylabel("Count")
plt.title("Price Range Distribution")
plt.show()

price_percentage = (price_range / price_range.sum()) * 100
print(price_percentage)

delivery_counts = df['Has Online delivery'].value_counts()
print(delivery_counts)

delivery_counts.plot(kind='pie', autopct='%1.1f%%')
plt.title("Online Delivery Availability")
plt.show()

delivery_percentage = (delivery_counts / delivery_counts.sum()) * 100
print(delivery_percentage)

avg_rating_delivery = df.groupby('Has Online delivery')['Aggregate rating'].mean()
print(avg_rating_delivery)

rating_distribution = df['Aggregate rating'].value_counts().sort_index()
print(rating_distribution)

rating_distribution.plot(kind='hist', bins=10)
plt.xlabel("Ratings")
plt.ylabel("Count")
plt.title("Rating Distribution")
plt.show()

most_common_rating = df['Aggregate rating'].mode()
print(most_common_rating)

avg_votes = df['Votes'].mean()
print(avg_votes)

cuisine_combo = df['Cuisines'].value_counts().head(10)
print(cuisine_combo)

cuisine_combo.plot(kind='bar')
plt.xlabel("Cuisine Combination")
plt.ylabel("Count")
plt.title("Top Cuisine Combinations")
plt.show()

cuisine_rating = df.groupby('Cuisines')['Aggregate rating'].mean()
cuisine_rating = cuisine_rating.sort_values(ascending=False)
print(cuisine_rating.head(10))

plt.scatter(df['Longitude'], df['Latitude'])
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Restaurant Locations")
plt.show()

city_location = df.groupby('City')[['Latitude', 'Longitude']].mean()
print(city_location.head(10))

restaurant_chains = df['Restaurant Name'].value_counts()
chains = restaurant_chains[restaurant_chains > 1]
print(chains.head(10))

chain_ratings = df.groupby('Restaurant Name')['Aggregate rating'].mean()
print(chain_ratings.sort_values(ascending=False).head(10))

df['Review_Length'] = df['Rating text'].astype(str).apply(len)

avg_review_length = df['Review_Length'].mean()
print(avg_review_length)

review_rating = df.groupby('Aggregate rating')['Review_Length'].mean()
print(review_rating)

review_rating.plot(kind='bar')
plt.xlabel("Aggregate Rating")
plt.ylabel("Average Review Length")
plt.title("Review Length vs Rating")
plt.show()

positive_reviews = df[df['Aggregate rating'] >= 3]['Rating text'].value_counts()
print(positive_reviews.head(10))

negative_reviews = df[df['Aggregate rating'] < 3]['Rating text'].value_counts()
print(negative_reviews.head(10))

positive_keywords = df[df['Aggregate rating'] >= 4]['Rating text'].value_counts()
negative_keywords = df[df['Aggregate rating'] < 3]['Rating text'].value_counts()

print(positive_keywords.head(10))
print(negative_keywords.head(10))

highest_votes = df.nlargest(5, 'Votes')
print(highest_votes[['Restaurant Name', 'Votes']])

lowest_votes = df.nsmallest(5, 'Votes')
print(lowest_votes[['Restaurant Name', 'Votes']])

correlation = df['Votes'].corr(df['Aggregate rating'])
print(correlation)

plt.scatter(df['Votes'], df['Aggregate rating'])
plt.xlabel("Votes")
plt.ylabel("Aggregate Rating")
plt.title("Votes vs Rating")
plt.show()

online_delivery = pd.crosstab(
    df['Price range'],
    df['Has Online delivery']
)

print(online_delivery)

online_delivery.plot(kind='bar')
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")
plt.title("Price Range vs Online Delivery")
plt.show()

table_booking = pd.crosstab(
    df['Price range'],
    df['Has Table booking']
)

print(table_booking)

table_booking.plot(kind='bar')
plt.xlabel("Price Range")
plt.ylabel("Number of Restaurants")
plt.title("Price Range vs Table Booking")
plt.show()

online_percentage = online_delivery.div(
    online_delivery.sum(axis=1), axis=0
) * 100
print(online_percentage)

booking_percentage = table_booking.div(
    table_booking.sum(axis=1), axis=0
) * 100
print(booking_percentage)
```
